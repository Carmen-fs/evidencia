```python
resumen = pd.read_csv(
    "Analisis GAC Full7agosto 2026(Resumen) (1).csv",
    header=5
)
```

```python
resumen.columns = resumen.columns.str.strip()
```

```python
resumen.columns.tolist()
```

```python
resumen_limpio = resumen.copy()
```

```python
meses_validos = [
    "Enero", "Febrero", "Marzo", "Abril",
    "Mayo", "Junio", "Julio", "Agosto",
    "Septiembre", "Octubre", "Noviembre", "Diciembre"
]

resumen_limpio = resumen_limpio[
    resumen_limpio["Mes"].astype(str).str.strip().isin(meses_validos)
].copy()
```

```python
print("Número de observaciones mensuales:", len(resumen_limpio))
```

```python
# Convertir SDC a numérica
resumen_limpio["SDC"] = pd.to_numeric(
    resumen_limpio["SDC"],
    errors="coerce"
)

# Quitar NaN únicamente para SDC
datos_sdc = resumen_limpio["SDC"].dropna()
```

```python
print("Datos usados:", len(datos_sdc))
print("NaN restantes:", datos_sdc.isna().sum())
```

```python
n_sdc = len(datos_sdc)

n_sdc
```

```python
Max_sdc = datos_sdc.max()
Min_sdc = datos_sdc.min()

Limites_sdc = [Min_sdc, Max_sdc]

Limites_sdc
```

```python
R_sdc = Max_sdc - Min_sdc

R_sdc
```

```python
ni_sdc = 1 + 3.32 * np.log10(n_sdc)

ni_sdc
```

```python
i_sdc = R_sdc / ni_sdc

i_sdc
```

```python
ni_sdc_redondeado = round(ni_sdc)

ni_sdc_redondeado
```

```python
intervalos_sdc = np.linspace(
    Min_sdc,
    Max_sdc,
    ni_sdc_redondeado + 1
)

intervalos_sdc
```

```python
categorias_sdc = [
    f"Muy bajo ({intervalos_sdc[0]:.0f} - {intervalos_sdc[1]:.0f})",
    f"Bajo ({intervalos_sdc[1]:.0f} - {intervalos_sdc[2]:.0f})",
    f"Medio-bajo ({intervalos_sdc[2]:.0f} - {intervalos_sdc[3]:.0f})",
    f"Medio ({intervalos_sdc[3]:.0f} - {intervalos_sdc[4]:.0f})",
    f"Medio-alto ({intervalos_sdc[4]:.0f} - {intervalos_sdc[5]:.0f})",
    f"Alto ({intervalos_sdc[5]:.0f} - {intervalos_sdc[6]:.0f})"
]

categorias_sdc
```

```python
sdc_categorizado = pd.cut(
    x=datos_sdc,
    bins=intervalos_sdc,
    labels=categorias_sdc,
    include_lowest=True
)

sdc_categorizado
```

```python
Tabla_freq_sdc = sdc_categorizado.value_counts().reset_index()

Tabla_freq_sdc
```

```python
Tabla_freq_sdc.columns = ["SDC", "count"]

Tabla_freq_sdc
```

```python
Filtro_sdc = Tabla_freq_sdc[
    Tabla_freq_sdc["count"] > 0
]

Filtro_sdc
```

```python
Filtro_index_sdc = Filtro_sdc.set_index("SDC")

Filtro_index_sdc
```

```python
orden_sdc = [
    categorias_sdc[0],  # Muy bajo
    categorias_sdc[1],  # Bajo
    categorias_sdc[2],  # Medio-bajo
    categorias_sdc[3],  # Medio
    categorias_sdc[4],  # Medio-alto
    categorias_sdc[5]   # Alto
]

Filtro_index_sdc = Filtro_index_sdc.reindex(orden_sdc)
```

```python
Filtro_index_sdc.plot(
    kind="bar",
    width=0.8,
    figsize=(10,4),
    color="pink",
    rot=45
)

plt.title("Categorías de SDC")
plt.xlabel("Rango de categorías")
plt.ylabel("Frecuencia")

plt.show()
```

```python
# Convertir PDM a numérica
resumen_limpio["PDM"] = pd.to_numeric(
    resumen_limpio["PDM"],
    errors="coerce"
)

# Quitar NaN únicamente de PDM
datos_pdm = resumen_limpio["PDM"].dropna()

print("Datos usados:", len(datos_pdm))
print("NaN restantes:", datos_pdm.isna().sum())
```

```python
# Número de datos
n_pdm = len(datos_pdm)

# Máximo y mínimo
Max_pdm = datos_pdm.max()
Min_pdm = datos_pdm.min()

# Rango
R_pdm = Max_pdm - Min_pdm

# Número de intervalos con Sturges
ni_pdm = 1 + 3.32 * np.log10(n_pdm)

# Ancho del intervalo
i_pdm = R_pdm / ni_pdm

print("n:", n_pdm)
print("Mínimo:", Min_pdm)
print("Máximo:", Max_pdm)
print("Rango:", R_pdm)
print("Intervalos según Sturges:", ni_pdm)
print("Ancho del intervalo:", i_pdm)
```

```python
ni_pdm_redondeado = round(ni_pdm)

intervalos_pdm = np.linspace(
    Min_pdm,
    Max_pdm,
    ni_pdm_redondeado + 1
)

intervalos_pdm
```

```python
niveles = [
    "Muy bajo",
    "Bajo",
    "Medio-bajo",
    "Medio",
    "Medio-alto",
    "Alto",
    "Muy alto"
]

categorias_pdm = []

for x in range(ni_pdm_redondeado):

    etiqueta = (
        f"{niveles[x]} "
        f"({intervalos_pdm[x]:.0f} - {intervalos_pdm[x+1]:.0f})"
    )

    categorias_pdm.append(etiqueta)

categorias_pdm
```

```python
pdm_categorizado = pd.cut(
    x=datos_pdm,
    bins=intervalos_pdm,
    labels=categorias_pdm,
    include_lowest=True
)

pdm_categorizado
```

```python
Tabla_freq_pdm = pdm_categorizado.value_counts().reset_index()

Tabla_freq_pdm.columns = ["PDM", "count"]

Tabla_freq_pdm
```

```python
Filtro_pdm = Tabla_freq_pdm[
    Tabla_freq_pdm["count"] > 0
]

Filtro_pdm
```

```python
Filtro_index_pdm = Filtro_pdm.set_index("PDM")

orden_pdm = categorias_pdm

Filtro_index_pdm = Filtro_index_pdm.reindex(orden_pdm).dropna()

Filtro_index_pdm
```

```python
Filtro_index_pdm.plot(
    kind="bar",
    width=0.8,
    figsize=(10,4),
    color="orange",
    rot=45
)

plt.title("Nivel de Pruebas de Manejo (PDM)")
plt.xlabel("Nivel de PDM")
plt.ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```
