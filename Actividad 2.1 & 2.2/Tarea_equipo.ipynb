```python
#Importamos las librerias pandas y numpy
import pandas as pd
import numpy as np
```

```python
#Leer archivo de Ventas por Asesor
ventas = pd.read_excel("Ventas_por_Asesor.xlsx")

ventas.head()
```

```python
#Información general
ventas.info()
```

```python
#Identificar valores nulos por columna
ventas.isnull().sum()
```

```python
#Convertir espacios vacíos en valores nulos
ventas = ventas.replace(r'^\s*$', np.nan, regex=True)
```

```python
ventas.isnull().sum()
```

```python
#Crear copia del DataFrame
ventas2 = ventas.copy()
```

```python
#Corregir los años de la fila correspondiente
#B3:L3 = 2024
ventas2.iloc[1, 1:12] = 2024

#M3:X3 = 2025
ventas2.iloc[1, 12:24] = 2025

#Y3:AE3 = 2026
ventas2.iloc[1, 24:31] = 2026

#Completar los años de la sección de acumulados/rankings

#AH2:AI2 = 2024
ventas2.iloc[0, 33:35] = 2024

#AJ2:AK2 = 2025
ventas2.iloc[0, 35:37] = 2025

#AL2:AM2 = 2026
ventas2.iloc[0, 37:39] = 2026

#Reemplazar todos los valores nulos de los meses por 0
ventas2.iloc[:, 1:31] = ventas2.iloc[:, 1:31].fillna(0)

print(ventas2.head())
```

```python
#Comprobar valores nulos
ventas2.isnull().sum()
```

```python
#Total de valores nulos
ventas2.isnull().sum().sum()
```

```python
#Crear archivo CSV final
ventas2.to_csv(
    "Ventas_por_Asesor_Sin_Nulos.csv",
    index=False
)
```

```python
#Leer archivo Principales Canales
canales = pd.read_excel("Principales_Canales.xlsx")
canales.head()
```

```python
#Identificar valores nulos por columna
canales.isnull().sum()
```

```python
#Crear copia del DataFrame
canales2 = canales.copy()
```

```python
#Completar bloques provenientes de celdas combinadas

#A5:AE5 = Piso
canales2.iloc[3, 0:31] = "Piso"

#A18:AE18 = Plazas
canales2.iloc[16, 0:31] = "Plazas"

#A33:AE33 = Digital
canales2.iloc[31, 0:31] = "Digital"

#A46:AE46 = Digital
canales2.iloc[44, 0:31] = "Digital"

#Completar títulos de indicadores en AG:AI
canales2.iloc[3, 32:35] = "Conversion"
canales2.iloc[7, 32:35] = "Aporte"
canales2.iloc[11, 32:35] = "Cota (Ventas) Ultimos 6ms"

canales2.iloc[16, 32:35] = "Conversion"
canales2.iloc[20, 32:35] = "Aporte"
canales2.iloc[26, 32:35] = "Cota (Ventas) Ultimos 6ms"

canales2.iloc[31, 32:35] = "Conversion"
canales2.iloc[35, 32:35] = "Aporte"
canales2.iloc[39, 32:35] = "Cota (Ventas) Ultimo Semestre"

#Reemplazar valores nulos por 0 en las columnas numéricas de meses
#Bloque Piso
canales2.iloc[5:14, 1:31] = (
    canales2.iloc[5:14, 1:31]
    .replace("-", 0)
    .fillna(0)
)

#Bloque Plazas
canales2.iloc[18:29, 1:31] = (
    canales2.iloc[18:29, 1:31]
    .replace("-", 0)
    .fillna(0)
)

#Primer bloque Digital
canales2.iloc[33:42, 1:31] = (
    canales2.iloc[33:42, 1:31]
    .replace("-", 0)
    .fillna(0)
)

#Segundo bloque Digital
canales2.iloc[46:55, 1:31] = (
    canales2.iloc[46:55, 1:31]
    .replace("-", 0)
    .fillna(0)
)
```

```python
#Comprobar valores nulos por columna
canales2.isnull().sum()
```

```python
#Comprobar total de valores nulos
canales2.isnull().sum().sum()
```

```python
#Crear archivo CSV final
canales2.to_csv(
    "Principales_Canales_Sin_Nulos.csv",
    index=False
)
```

```python
#En las columnas de la base de datos ventas por asesor, a los meses de 2024, 2025 y 2026, los valores nulos los reemplazé por 0 
# mediante el método fillna(0), debido a que si no había información en estas variables se entendió como que no hbuo 
# ventas registradas para ese periodo. Además, se corrigieron manualmente los valores correspondientes 
# a los años 2024, 2025 y 2026 en las filas utilizadas como encabezados de los diferentes periodos y en las secciones 
# de acumulados y rankings. Finalmente, se verificó nuevamente la cantidad de valores nulos con isnull().sum() y isnull().sum().sum(). 
# Una vez realizado esto, el DataFrame limpio se exportó como Ventas_por_Asesor_Sin_Nulos.csv

#En las columnas de la base de datos de principales canales, se completaron manualmente los valores de las categorías Piso, Plazas 
# y Digital, manteniendo la estructura y clasificación original de los datos. De igual manera, se completaron los títulos 
# correspondientes a los indicadores Conversión, Aporte y Cota de Ventas de los últimos seis meses o semestre, según cada sección. 
# Para las columnas numéricas correspondientes a los meses, se utilizaron dos técnicas de limpieza. Primero, los valores 
# representados mediante un guion ("-") fueron sustituidos por 0 utilizando replace("-", 0). Posteriormente, los valores 
# nulos restantes fueron reemplazados mediante fillna(0). Este procedimiento se aplicó por separado en los bloques correspondientes 
# a Piso, Plazas y Digital, considerando que un dato vacío dentro de estas variables numéricas representa la ausencia de un valor 
# registrado. Finalmente, se comprobó nuevamente la cantidad de valores nulos por columna y en todo el dataset. Después del 
# tratamiento, el archivo limpio se exportó con el nombre Principales_Canales_Sin_Nulos.csv.
```
