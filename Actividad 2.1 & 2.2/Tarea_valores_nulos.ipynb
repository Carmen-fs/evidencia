```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("Citas_Digital_Filtrado.csv")

data.head()
```

```python
#Información general del DataFrame
data.info()
```

```python
#EMPEZAMOS A IDENTIFICAR LOS DATOS FALTANTES
```

```python
#Identificar valores nulos por columna
valores_nulos = data.isnull().sum()

valores_nulos
```

```python
#Identificar valores nulos de todo el DataFrame
valores_nulos_total = data.isnull().sum().sum()

print(valores_nulos_total)
```

```python
#Realizamos una copia del DataFrame
data2 = data.copy()

print(data2)
```

```python
#Sustituir valores nulos de BDC con la moda
data2["BDC "] = data2["BDC "].fillna(
    data2["BDC "].mode()[0]
)
```

```python
#Sustituir valores nulos de Fecha
data2["Fecha"] = data2["Fecha"].fillna("Sin fecha")
```

```python
#Sustituir valores nulos de Nombre Cliente
data2["Nombre Cliente "] = data2["Nombre Cliente "].fillna(
    "No registrado"
)
```

```python
#Sustituir valores nulos de Telefono
data2["Telefono "] = data2["Telefono "].fillna(
    "No registrado"
)
```

```python
#Sustituir valores nulos de Estatus de Lead con la moda
data2["Estatus de Lead"] = data2["Estatus de Lead"].fillna(
    data2["Estatus de Lead"].mode()[0]
)
```

```python
#Limpiar algunos valores incorrectos de Potencial de compra
data2["Potencial de compra"] = (
    data2["Potencial de compra"]
    .astype(str)
    .str.strip()
    .replace({
        "80%%": "0.8",
        "80&": "0.8",
        "8'%": "0.8",
        "Ventas": np.nan,
        "nan": np.nan
    })
)
```

```python
#Convertir Potencial de compra a tipo numérico
data2["Potencial de compra"] = pd.to_numeric(
    data2["Potencial de compra"],
    errors="coerce"
)
```

```python
#Sustituir valores nulos de Potencial de compra con la mediana
data2["Potencial de compra"] = data2["Potencial de compra"].fillna(
    data2["Potencial de compra"].median()
)
```

```python
#Sustituir valores nulos de PDM con la moda
data2["PDM"] = data2["PDM"].fillna(
    data2["PDM"].mode()[0]
)
```

```python
#Sustituir valores nulos de SDC con la moda
data2["SDC"] = data2["SDC"].fillna(
    data2["SDC"].mode()[0]
)
```

```python
#Sustituir valores nulos de Venta con la moda
data2["Venta"] = data2["Venta"].fillna(
    data2["Venta"].mode()[0]
)
```

```python
#Sustituir valores nulos de Asesor Asignado con la moda
data2["Asesor Asignado"] = data2["Asesor Asignado"].fillna(
    data2["Asesor Asignado"].mode()[0]
)
```

```python
#Sustituir valores nulos de preguntas abiertas
data2["¿Por que no ha visitado la agencia?"] = data2[
    "¿Por que no ha visitado la agencia?"
].fillna("Sin información")
```

```python
data2["¿Por que no hay solicitud de credito?"] = data2[
    "¿Por que no hay solicitud de credito?"
].fillna("Sin información")
```

```python
data2["¿Por que no hubo prueba de manejo ?"] = data2[
    "¿Por que no hubo prueba de manejo ?"
].fillna("Sin información")
```

```python
data2["¿Por que no hubo proceso wow?"] = data2[
    "¿Por que no hubo proceso wow?"
].fillna("Sin información")
```

```python
data2["¿Por que se asigno antes de vistar piso?"] = data2[
    "¿Por que se asigno antes de vistar piso?"
].fillna("Sin información")
```

```python
#Sustituir valores nulos de la columna de observaciones
data2["Unnamed: 15"] = data2["Unnamed: 15"].fillna(
    "Sin observaciones"
)
```

```python
#Corroboramos valores nulos por columna
valores_nulos = data2.isnull().sum()

print(valores_nulos)
```

```python
#Corroboramos el total de valores nulos
valores_nulos_total = data2.isnull().sum().sum()

print(valores_nulos_total)
```

```python
#Convertir DataFrame sin valores nulos a CSV
data2.to_csv(
    "Citas_Digital_Sin_Nulos.csv",
    index=False
)
```
